```python
import pandas as pd
from youtube_comment_downloader import YoutubeCommentDownloader, SORT_BY_POPULAR
import demoji

# 1. Initialize emoji database
try:
    demoji.download_codes()
except:
    pass

# --- LOW STRESS KEYWORDS ONLY ---
low_keywords = [
    "calm", "relaxing", "peaceful", "soothing", "comforting", "zen", "tranquil", 
    "interesting", "informative", "fascinating", "educational", "insightful",
    "learned", "knowledge", "science", "fact", "brain", "understanding",
    "thanks", "thank you", "grateful", "appreciate", "helpful", "wonderful",
    "clear", "focused", "balanced", "stable", "nice", "amazing", "curiosity","motivated"
]

# 2. Scraper Settings
downloader = YoutubeCommentDownloader()
urls = ['https://www.youtube.com/watch?v=grBFMP3HDZA',
       'https://www.youtube.com/watch?v=1ex_bNIFR1A',
        'https://www.youtube.com/watch?v=QJQa2NGK9UY',
        'https://www.youtube.com/watch?v=p0RR4LWlt5w',
        'https://www.youtube.com/watch?v=1jLTWm63WNE',
        'https://www.youtube.com/watch?v=ioVPxHX-gUM',
        'https://www.youtube.com/watch?v=5Q2Pc-e-8Qc',
        'https://www.youtube.com/watch?v=DXT9dF-WK-I',
        'https://www.youtube.com/watch?v=MdxH1x1u0tk',
        'https://www.youtube.com/watch?v=mZBwLwlQg2s',
        'https://www.youtube.com/watch?v=jp2M1hIEtsk',
        'https://www.youtube.com/watch?v=Kr7eiJz0kvc',
        'https://www.youtube.com/watch?v=sUwD3GRPJos',
        'https://www.youtube.com/watch?v=7JPTlqRRf1w',
        'https://www.youtube.com/watch?v=5Q2Pc-e-8Qc',
        'https://www.youtube.com/watch?v=x0qcqPcEfoc',
        'https://www.youtube.com/watch?v=wLXFynJCoF4'
        
        
       ]
all_comments = []

print("Starting Extraction for LOW STRESS comments (with Emoji Removal)...")

for url in urls:
    try:
        generator = downloader.get_comments_from_url(url, sort_by=SORT_BY_POPULAR)
        count = 0
        
        for comment in generator:
            raw_text = comment['text']
            
            # الخطوة دي بتمسح الإيموجي تماماً من النص
            clean_text = demoji.replace(raw_text, "")
            clean_text = " ".join(clean_text.split())
            
            if clean_text:
                text_lower = clean_text.lower()
                
                # بنشوف هل الكومنت فيه أي كلمة من الـ Low Keywords؟
                is_low_stress = any(word in text_lower for word in low_keywords)
                
                if is_low_stress:
                    all_comments.append({
                        'video_url': url,
                        'original_text': raw_text, # النص الأصلي بالإيموجي (للمراجعة)
                        'clean_text': clean_text,     # النص النضيف اللي هتستخدميه في الـ Model
                        'label': 'low'
                    })
                    count += 1
                    
                    if count % 100 == 0:
                        print(f"Validated {count} low-stress comments...")

        print(f"✅ Successfully finished. Total relevant comments: {count}")

    except Exception as e:
        print(f"❌ Error: {e}")

# 3. Create DataFrame and Save File
if all_comments:
    df = pd.DataFrame(all_comments)
    df.to_csv('youtube_low_stress_dataset.csv', index=False, encoding='utf-8-sig')
    print(f"\nFile saved: youtube_low_stress_dataset.csv")
    print(f"Final Count: {len(df)} rows.")
    display(df.head(10))
else:
    print("No comments matched your keywords!")
```
