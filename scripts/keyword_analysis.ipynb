```python
import pandas as pd
from collections import Counter
import re
```

```python
def better_keyword_analysis(csv_path):
    df = pd.read_csv(csv_path)
    
    # Extended stopwords list
    stopwords = set([
        'i', 'me', 'my', 'the', 'a', 'an', 'and', 'or', 'but',
        'in', 'on', 'at', 'to', 'for', 'of', 'with', 'is', 'it',
        'this', 'that', 'was', 'are', 'be', 'have', 'had', 'has',
        'do', 'did', 'not', 'he', 'she', 'they', 'we', 'you',
        'so', 'if', 'just', 'like', 'get', 'got', 'can', 'will',
        'about', 'dont', 'know', 'been', 'what', 'when', 'really',
        'because', 'want', 'would', 'from', 'some', 'time', 'feel',
        'there', 'their', 'said', 'more', 'also', 'than', 'then',
        "don't", "i've", "it's", "i'm", "i'll", 'your', 'our',
        'very', 'even', 'much', 'only', 'them', 'been', 'who',
        'which', 'after', 'before', 'could', 'should', 'would',
        'one', 'two', 'all', 'any', 'how', 'now', 'his', 'her'
    ])
    
    print("🔑 Most DISTINCTIVE keywords per stress level:")
    print("="*50)
    
    for label in ['high', 'medium', 'low']:
        texts = ' '.join(
            df[df['stress_label'] == label]['full_text'].tolist()
        )
        words = [
            w.lower().strip(".,!?'\"") 
            for w in texts.split() 
            if w.lower().strip(".,!?'\"") not in stopwords 
            and len(w) > 4
        ]
        top_words = Counter(words).most_common(15)
        print(f"\n{label.upper()} STRESS:")
        print([w for w, c in top_words])
```

```python
if __name__ == "__main__":
    csv_path = r"F:\UNIVERSITY\NU 3\semester 2\Machine Intelligence\Project\StressClassifier\data\parsed\verified_clean_data.csv"
    better_keyword_analysis(csv_path)
```
