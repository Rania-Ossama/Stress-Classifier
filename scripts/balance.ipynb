```python
import pandas as pd
```

```python
def balance_dataset(csv_path, output_folder, samples_per_class=10000):
    df = pd.read_csv(csv_path)
    
    print(f"Before balancing:\n{df['stress_label'].value_counts()}\n")
    
    # Sample equally from each class
    balanced = df.groupby('stress_label').apply(
        lambda x: x.sample(
            min(len(x), samples_per_class), 
            random_state=42
        )
    ).reset_index(drop=True)
    
    # Shuffle the dataset
    balanced = balanced.sample(frac=1, random_state=42).reset_index(drop=True)
    
    print(f"After balancing:\n{balanced['stress_label'].value_counts()}\n")
    print(f"✅ Total samples: {len(balanced)}")
    
    # Save
    output_path = f"{output_folder}/balanced_data.csv"
    balanced.to_csv(output_path, index=False, encoding='utf-8')
    print(f"💾 Saved to {output_path}")
    
    return balanced
```
